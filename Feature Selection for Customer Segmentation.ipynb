```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
```

```python
#check missing value
df.isnull().sum()
```

```python
#Handle Missing Values
numerical_cols = df.select_dtypes(include=['int64','float64']).columns

for col in numerical_cols:
    df[col].fillna(df[col].median(),inplace=True)
```

```python
categorical_cols = df.select_dtypes(include=['object','category','bool']).columns

for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
#Verify Missing Values
print("Missing Values After Cleaning")
print(df.isnull().sum())
```

```python
#Remove Duplicate 
print("Duplicate Rows :", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Shape After Removing Duplicates :", df.shape)
```

```python
#Remove Non-Informative Columns
columns_to_remove = [
    'customer_id',
    'month',
    'Month_Name'
]

df.drop(columns=columns_to_remove, inplace=True, errors='ignore')
```

```python
#Select Features for Segmentation
selected_features = [

    'monthly_transactions_avg',
    'avg_monthly_balance',
    'num_products_held',
    'digital_adoption_score',
    'preferred_channel',
    'credit_card_issued',
    'loan_taken',
    'Active customer'
]

segmentation_df = df[selected_features].copy()
```

```python
print(segmentation_df.head())
```

```python
#Encode Categorical Variables
categorical_features = [

    'preferred_channel',
    'credit_card_issued',
    'loan_taken',
    'Active customer'
]

segmentation_df = pd.get_dummies(

    segmentation_df,

    columns=categorical_features,

    drop_first=True
)
```

```python
#Verify Data Types
print(segmentation_df.dtypes)
```

```python
#Scale Features
scaler = StandardScaler()

segmentation_scaled = scaler.fit_transform(segmentation_df)
```

```python
#Convert Back to DataFrame
segmentation_scaled = pd.DataFrame(

    segmentation_scaled,

    columns=segmentation_df.columns
)
```
